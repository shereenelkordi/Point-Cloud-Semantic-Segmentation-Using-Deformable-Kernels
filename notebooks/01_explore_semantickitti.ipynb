# Exploring SemanticKITTI
Counts the scans, images and labels in each sequence, defines the train/val/test split, and reads a raw Velodyne scan and its labels.

```python
import sys
import os
import json
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import Dataset, DataLoader
import time
import glob
```

```python
root = os.path.join(os.environ.get("SEMKITTI_ROOT", "../dataset/sequences/"), "")
```

```python
sequences = sorted(glob.glob(root+"*"))
print(len(sequences))
```

```python
print(sequences[0])
```

```python
#number of PC in each sequence
count_pc = []
count_image2 = []
count_image3 = []
count_labels = []
for seq in sequences:
    count = len(glob.glob(seq+"/velodyne/*"))
    count_pc.append(count)
    count = len(glob.glob(seq+"/image_2/*"))
    count_image2.append(count)
    count = len(glob.glob(seq+"/image_3/*"))
    count_image3.append(count)
    count = len(glob.glob(seq+"/labels/*"))
    count_labels.append(count)
    
print(f"num of pc files in each sequence = {count_pc}")
print(f"num of image_2 files in each sequence = {count_image2}")
print(f"num of image_3 files in each sequence = {count_image3}")
print(f"num of labels files in each sequence = {count_labels}")
```
